```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("planets.csv")
print("Dataset loaded successfully!")
print("Shape:",df.shape)
```

```python
df.columns=(df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","_")
           )
print(df.columns.tolist())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col]<lower)|(df[col]>upper)]
    print(col, ":",len(outliers), "possible outliers")
```

```python
mass_col = [c for c in df.columns if "mass" in c] 
[0]
plt.figure(figsize=(10,5))
plt.bar(df.iloc[:, 0].values, df[mass_col 
    [0]].values)

plt.yscale("log")
plt.xlabel("Planet")
plt.ylabel("Mass")
plt.title("Planetary Mass Comparison")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
diameter_col = [c for c in df.columns if "diameter" in c][0]
plt.figure(figsize=(8,5))
plt.scatter(df[diameter_col],df[mass_col])
for index, row in df.iterrows():
    plt.text(
        x=row["diameter"],
        y=row["mass"],
        s=row.iloc[0],
        fontdict = dict(color="black", size=10))
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Diameter")
plt.ylabel("Mass")
plt.title("Planetary Diamter vs Mass")
plt.tight_layout()
plt.show()
```

```python
distance_col = [c for c in df.columns if "semi_major" in c or "distance" in c][0]
period_col = [c for c in df.columns if "orbital_period" in c][0]
plt.figure(figsize=(8,5))
plt.scatter(df[distance_col],df[period_col])
for index, row in df.iterrows():
    plt.text(
        x=row["distance_from_sun"],
        y=row["orbital_period"],
        s=row.iloc[0],
        fontdict = dict(color="black", size=10))
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Distance from the Sun")
plt.ylabel("Orbital Period")
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[numeric_cols] = scaler.fit_transform(df[numeric_cols]
                                              )
df_scaled.head()
```
